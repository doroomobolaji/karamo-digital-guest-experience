```python
## Typescript 
# Use these interfaces to type responses from the FastAPI endpoints cleanly in TypeScript / React / Next.js

export type CancellationRiskTier = "LOW" | "MEDIUM" | "HIGH";

export type GuestSegmentLabel =
  | "SHORT_STAY_VALUE"
  | "LONG_STAY_BUDGET"
  | "HIGH_SPEND_EXTENDED";

export type BookingStatus = "Completed" | "Confirmed" | "Cancelled";
export type PaymentStatus = "Paid" | "Refunded";
export type PaymentMethod = "Mobile Money" | "Card" | "USSD" | "Bank Transfer";

export interface StaySummary {
  room_type: string;
  room_id: string; // Kept as string per Cleaning_Log
  checkin_date: string; // ISO date YYYY-MM-DD
  checkout_date: string; // ISO date YYYY-MM-DD
  stay_nights: number;
  guest_count: number;
  lead_time_days: number;
}

export interface FinancialSummary {
  amount_per_night: number;
  total_amount: number;
  currency: string | null; // Null until confirmed in Cleaning_Log
  payment_method: PaymentMethod | string;
  payment_reference: string;
}

export interface CancellationRisk {
  cancellation_risk_score: number;
  cancellation_risk_tier: CancellationRiskTier;
  model_version: string;
  is_directional_estimate: boolean;
  recommended_action: string;
}

export interface BookingInsightsResponse {
  booking_id: string;
  user_id: string;
  booking_status: BookingStatus | string;
  payment_status: PaymentStatus | string;
  stay_summary: StaySummary;
  financials: FinancialSummary;
  risk: CancellationRisk;
}

export interface GuestSegmentInfo {
  segment_id: number;
  segment_label: GuestSegmentLabel;
  segment_version: string;
  preferred_room_type: string;
  preferred_payment_method: string;
  avg_stay_nights: number;
  avg_total_amount: number;
}

export interface UiPersonalization {
  show_repeat_guest_badge: boolean;
  featured_offer_type: string;
}

export interface GuestProfileResponse {
  user_id: string;
  is_repeat_guest: boolean;
  total_bookings: number;
  segment: GuestSegmentInfo;
  ui_personalization: UiPersonalization;
}

export interface CheckoutMethodMetric {
  payment_method: PaymentMethod | string;
  booking_count: number;
  avg_booking_duration_min: number;
  avg_payment_duration_min: number;
  observed_cancellation_rate: number;
  high_friction_flag: boolean;
}

export interface CheckoutHealthResponse {
  as_of: string; // ISO timestamp
  is_directional_estimate: boolean;
  methods: CheckoutMethodMetric[];
}

export type AsyncUiState<T> =
  | { status: "idle" }
  | { status: "loading" }
  | { status: "success"; data: T }
  | { status: "empty"; message: string }
  | { status: "error"; statusCode?: number; message: string };
```
